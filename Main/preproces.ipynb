```python
import pandas as pd
import openpyxl
```

```python
df = pd.read_csv("EDGAR 3.5 JLU Giessen (1).csv", encoding='utf-8')
print(len(df.columns))
df = df.replace(' - , -' , '-,-,-') 

df = df.loc[:, ~df.columns.str.contains('plasmid', case=False)]
for col in df.columns:
    df[col]=df[col].str.replace(', ,', ',')

df.head(100)
```

```python
df = df.applymap(lambda x: x.split(',')[0] if isinstance(x, str) else x)
df.replace('-', pd.NA, inplace=True)
df.to_csv("cleaned_EDGAR.csv", index=False)
df
```

```python
# Collect unique non-null gene names per strain (column)
unique_genes = df.apply(lambda col: list(col.dropna().unique()))

# Convert to JS object
js_map = "const geneCounts = {\n"
for strain, genes in unique_genes.items():
    gene_list = ', '.join(f'"{gene}"' for gene in genes)
    js_map += f'  "{strain}": [{gene_list}],\n'
js_map += "};"

# Save to JS file
with open("gene_counts.js", "w") as f:
    f.write(js_map)
```
